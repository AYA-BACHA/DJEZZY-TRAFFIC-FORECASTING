# Djezzy Traffic Forecasting — 02 Exploratory Data Analysis (EDA)

Analysis of temporal trends, technology distributions, Algerian cultural impacts, and congestion dynamics.

```python
import pandas as pd
import plotly.express as px

df = pd.read_parquet('../data/processed/djezzy_cleaned.parquet')
print('Loaded rows:', len(df))
df.head()
```

## 1. Traffic by Technology Generation

```python
df.groupby('technology')['dl_traffic_volume_gb'].agg(['count', 'mean', 'sum', 'max'])
```

## 2. Diurnal & Weekly Patterns (Hour vs Day of Week)

```python
df['hour'] = df['timestamp'].dt.hour
df['day_of_week'] = df['timestamp'].dt.dayofweek
pivot_4g = df[df['technology'] == '4G'].groupby(['day_of_week', 'hour'])['dl_traffic_volume_gb'].mean().unstack()
px.imshow(pivot_4g, labels=dict(x='Hour of Day', y='Day of Week (0=Mon, 4=Fri, 5=Sat)'), title='4G Average DL Traffic (GB/h)')
```

## 3. Top Wilayas by Traffic

```python
df.groupby('wilaya_name')['dl_traffic_volume_gb'].mean().sort_values(ascending=False).head(10)
```
